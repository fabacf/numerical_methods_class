## Exercises 1.26, 1.27

```python
import numpy as np
import matplotlib.pyplot as plt
import math as m

from pygments.lexers import factor

x = np.linspace(-1.5, 1.5, 100)

# Exercise 1.26 c)

f = np.exp(x)
g1 = np.ones_like(x)
g2 = 1 + x

# Exercise 1.27 a) iii.

g3 = 1 + x + (0.5 * (x ** 2))

# Exercise 1.27 b) iii.

g4 = 1 + x + 0.5 * (x ** 2) + ((1/6) * x**3)

# Exercise 1.27 c)

g5 = 1 + x + 0.5 * (x ** 2) + ((1/6) * x**3) + ((1/24) * x**4)

g6 = 1 + x + 0.5 * (x ** 2) + ((1/6) * x**3) + ((1/24) * x**4) + ((1/120) * x**5)

plt.figure(figsize=(10, 7))
plt.plot(x, f, 'r--', linewidth=2.5, label='f(x) = e^x')
plt.plot(x, g1, 'b-.', linewidth=2.5, label='constant')
plt.plot(x, g2, 'k-', linewidth=2.5, label='linear')
plt.plot(x, g3, 'y-', linewidth=2.5, label= 'g3')
plt.plot(x, g4, 'o-', linewidth=2.5, label= 'g4')
plt.plot(x, g5, 'yx', linewidth=2.5, label= 'g5')
plt.plot(x, g6, 'ro', linewidth=2.5, label= 'g6')

plt.title('Two approximations of e^x at x=0')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```
