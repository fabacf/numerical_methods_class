```python
def bisection(f, a, b, nah = 0.001, max_iter=100):
    # vorzeichen prüfen Produkt von f(a) und f(b)
    if f(a) * f(b) > 0:
        raise ValueError("f(a) und f(b) haben gleiches Vorzeichen")

    for i in range(max_iter):
        m = (a + b) / 2
        fm = f(m)

        # chechken wie nah m an 0 ist
        if abs(fm) < nah:
            return m


        if f(a) * fm < 0:      # vzw zwischen a und m
            b = m
        else:                  # vzw zwischen m und b
            a = m

    return (a + b) / 2
```

```python
import numpy as np
import matplotlib.pyplot as plt

# a) x^2 - 2 = 0
fa = lambda x: x**2 - 2

# b) sin(x) + x^2 = 2 ln(x) + 5
fb = lambda x: np.sin(x) + x**2 - 2*np.log(x) - 5

# c) (5 - x) e^x - 5 = 0
fc = lambda x: (5 - x) * np.exp(x) - 5

x = np.linspace(0.01, 5, 500)
for f, name in [(fa, "a"), (fb, "b"), (fc, "c")]:
    plt.plot(x, f(x), label=name)
plt.axhline(0, color="black")
plt.ylim(-10, 20)
plt.legend()
plt.grid()
plt.show()

print(round(bisection(fa, 0, 2),3))
print(round(bisection(fb, 1, 5),3))
print(round(bisection(fc, 0, 5),3))
```
