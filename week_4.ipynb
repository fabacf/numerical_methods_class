```python
def newton(f, fd, x, e = 0.0001):
    #lamba funktion als eingabe, die ableitung davon, einen ersten guess und optional error toleranz

    while abs(f(x)) > e:
        #steigung m von f strich xn berechnen
        m = fd(x)
        # x n+1 ist: xn - funktionswert von xn geteilt durch m
        x = x - (f(x)/m)
    return x


```

```python
fa = lambda x: x**2 - 2
fd_a = lambda x: 2*x

fb = lambda x: 2* x**3 + 5
fd_b = lambda x: 6*x**2

fc = lambda x: 0.5* x**2 - 2
fd_c = lambda x: x
```

```python
newton(fa,fd_a, 5)
```

```python
newton(fb, fd_b, 3)
```

```python
newton(fc, fd_c, 4)
```
