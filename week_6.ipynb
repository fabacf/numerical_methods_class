# Exercise 3.32

```python
import numpy as np


def riemann_sums(f, lb, ub, ns, p = "left"):

    x = np.linspace(lb, ub, ns + 1)
    step = (ub - lb) / ns

    if p == "left":
        i = x[:-1]
    elif p == "right":
        i = x[1:]
    elif p == "midpoint":
        i = x[:-1] + step/2

    return np.sum(f(i) * step)
```

```python
f = lambda x: x**2

riemann_sums(f, 0, 2, 4, "left")
```

# Exericse 3.33 b)

```python
f2 = lambda x: np.sin(x)

riemann_sums(f2, 0, 1, 2, "left")
```

```python
riemann_sums(f2, 0, 1, 4, "left")
```

```python
exact = 1 - np.cos(1)

for k in range(1, 9):
    j = 2**-k
    n = int(1 / j)
    approx = riemann_sums(f2, 0, 1, n)
    error = abs(approx - exact) / exact * 100
    print(j, approx, exact, error)
```
