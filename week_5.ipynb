```python
import numpy as np

f = lambda x: np.sin(x) * (1 - x)

exact = np.sin(-1)

for i in range(1, 11):
    h = 2**-i
    approx = (f(1 + h) - f(1)) / h
    err = 18.54
    print(f"{h}, {approx}, {exact}, {err}%")
```

## Exercise 3.14

```python
import numpy as np
import matplotlib.pyplot as plt

f =lambda x:np.sin(x)* (1-x) # defines the function f

exact =-np.sin(1) #calculates the exact value of the function on x = 1, in this case sin(-1) * (1-1)

H = 2.0**(-np.arange(1,10)) #whatdoesthislinedo?

AbsPctError= [] #start off with a blank list of errors
for h in H:
    approx = (f(1 + h) - f(1)) / h
    AbsPctError.append(np.abs((approx-exact)/exact))
    if h==H[0]:
        print("h=",h,"\tAbsolutePctError=",AbsPctError[-1])
    else:
        err_reduction_factor = AbsPctError[-2]/AbsPctError[-1]
        print("h=",h,"\tAbsolutePctError=",AbsPctError[-1], "with error reduction",err_reduction_factor)

plt.loglog(H,AbsPctError,'b-*') #Why are we build aloglog plot? To show the error reduction
plt.grid()
plt.show()
```
